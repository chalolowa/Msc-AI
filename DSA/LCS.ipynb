## Longest Common Subsequence (LCS) function implementation

```python
def lcs(s1, s2):
    m, n = len(s1), len(s2)

    # Create 2D table dp
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    # Fill the table
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[m][n]
```

## lets introduce test samples

```python
#short string sample
s1 = "ABCD"
s2 = "ACBD"
print("LCS length (short):", lcs(s1, s2))
```
